```python
# loading the library
import pandas as pd

train = pd.read_csv('C:\\Users\\CDT WO AZEEZ SAMAD\\Downloads\\climate-health-prediction\\data\\Train.csv')
test = pd.read_csv('C:\\Users\\CDT WO AZEEZ SAMAD\\Downloads\\climate-health-prediction\\data\\Test.csv')
features = pd.read_csv('C:\\Users\\CDT WO AZEEZ SAMAD\\Downloads\\climate-health-prediction\\data\\climate_features.csv')
```

```python
#checking the shape of the dataframes
print("Train shape:", train.shape)
print("Test shape:", test.shape)
print("Features shape:", features.shape)
```

```python
#displaying the first few rows of the dataframes
print("\nTrain DataFrame:")
display(train.head())
print("\nTest DataFrame:")
display(test.head())
print("\nFeatures DataFrame:")
display(features.head())
```

```python
#checking the column names of the dataframes
print("Train Columns")
display(train.columns.tolist())
print("Test Columns")
display(test.columns.tolist())
print("Features Columns")
display(features.columns.tolist())
```

```python
#checking for duplicates in the dataframes
print("Train Duplicates:", train.duplicated("ID").sum())
print("Test Duplicates:", test.duplicated("ID").sum())
print("Features Duplicates:", features.duplicated("ID").sum())
```

```python
#  Check missing values
print("\nMissing values in Train:")
display(train.isnull().sum().sort_values(ascending=False).head(15))

print("\nMissing values in Test:")
display(test.isnull().sum().sort_values(ascending=False).head(15))

print("\nMissing values in Features:")
display(features.isnull().sum().sort_values(ascending=False).head(15))
```

```python
# investigating target variable distribution in the train dataset
print("\nTarget variable distribution in Train:")
display(train["is_climate_sensitive"].value_counts())

print("\nTarget proportions:")
display(train["is_climate_sensitive"].value_counts(normalize=True))
```

```python

# Check whether every train ID exists in features
train_ids_match = train["ID"].isin(features["ID"]).all()

# Check whether every test ID exists in features
test_ids_match = test["ID"].isin(features["ID"]).all()

# Check whether every feature ID belongs to train or test
all_data_ids = pd.concat([train["ID"], test["ID"]])
extra_feature_ids = ~features["ID"].isin(all_data_ids)

print("All train IDs found in features:", train_ids_match)
print("All test IDs found in features:", test_ids_match)
print("Extra IDs in features:", extra_feature_ids.sum())

# Check whether feature IDs cover the combined dataset exactly
print("Combined train/test unique IDs:", all_data_ids.nunique())
print("Unique feature IDs:", features["ID"].nunique())
print("Do feature IDs cover the combined dataset exactly?", features["ID"].nunique() == all_data_ids.nunique())
```

```python

# 1. Remove deathdate from the climate features
# because train and test already contain this column
features_to_merge = features.drop(columns=["deathdate"])

# 2. Merge climate features into training data
train_merged = train.merge(
    features_to_merge,
    on="ID",
    how="left",
    validate="one_to_one"
)

# 3. Merge climate features into test data
test_merged = test.merge(
    features_to_merge,
    on="ID",
    how="left",
    validate="one_to_one"
)

# 4. Inspect the results
print("Original train shape:", train.shape)
print("Merged train shape:", train_merged.shape)

print("\nOriginal test shape:", test.shape)
print("Merged test shape:", test_merged.shape)

print("\nMissing values in merged train:",
      train_merged.isnull().sum().sum())

print("Missing values in merged test:",
      test_merged.isnull().sum().sum())
```

```python

# Display every merged column and its data type
print("TRAINING DATA COLUMNS AND TYPES")
train_merged.info()

print("\nCOLUMN NAMES")
print(train_merged.columns.tolist())

print("\nFIRST 5 MERGED RECORDS")
display(train_merged.head())

print("\nUNIQUE VALUES IN CATEGORICAL COLUMNS")
for column in ["zone", "gender", "location"]:
    print(f"\n{column}:")
    print(train_merged[column].value_counts().head(10))
```

```python

# 1. Count unique values in categorical columns
for column in ["zone", "gender", "location"]:
    print(f"{column}: {train_merged[column].nunique()} unique categories")
    print(train_merged[column].value_counts().head(10))
    print()

# 2. Inspect the date range
dates = pd.to_datetime(
    train_merged["deathdate"],
    format="mixed",
    dayfirst=False,
    errors="coerce"
)

print("Invalid dates:", dates.isna().sum())
print("Earliest date:", dates.min())
print("Latest date:", dates.max())

# 3. Inspect numerical feature ranges
numeric_columns = train_merged.select_dtypes(
    include="number"
).columns.drop("is_climate_sensitive")

display(train_merged[numeric_columns].describe().T)
```
